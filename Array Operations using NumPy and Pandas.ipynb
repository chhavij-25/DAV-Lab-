```python
import numpy as np
import pandas as pd

# ============================================================
# 1. ONE-DIMENSIONAL ARRAY
# ============================================================

print("=" * 60)
print("1. ONE-DIMENSIONAL ARRAY")
print("=" * 60)

arr = np.array([10, 20, 30, 40, 50, 60, 70])

print("Original Array:", arr)

# Indexing
print("\n--- Indexing Operations ---")
print("First Element:", arr[0])
print("Last Element:", arr[-1])
print("Third Element:", arr[2])
print("Second Last Element:", arr[-2])

# Slicing
print("\n--- Slicing Operations ---")
print("Elements from Index 1 to 4:", arr[1:5])
print("First Three Elements:", arr[:3])
print("Elements from Index 2:", arr[2:])
print("Every Second Element:", arr[::2])
print("Every Third Element:", arr[::3])
print("Reversed Array:", arr[::-1])
print("Index 1 to 5 with Step 2:", arr[1:6:2])

# ============================================================
# 2. TWO-DIMENSIONAL ARRAY
# ============================================================

print("\n" + "=" * 60)
print("2. TWO-DIMENSIONAL ARRAY")
print("=" * 60)

arr2 = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Original 2D Array:\n", arr2)

# Indexing
print("\n--- 2D Indexing Operations ---")
print("First Row:", arr2[0])
print("First Element:", arr2[0, 0])
print("Element at Row 2, Column 3:", arr2[1, 2])
print("Last Element:", arr2[-1, -1])

# Slicing
print("\n--- 2D Slicing Operations ---")
print("First Two Rows:\n", arr2[:2])
print("First Two Columns:\n", arr2[:, :2])
print("First Two Rows and Columns:\n", arr2[:2, :2])
print("Every Second Row:\n", arr2[::2])
print("Every Second Column:\n", arr2[:, ::2])
print("Reversed Rows:\n", arr2[::-1])
print("Reversed Columns:\n", arr2[:, ::-1])

# ============================================================
# 3. NUMPY OPERATIONS
# ============================================================

print("\n" + "=" * 60)
print("3. NUMPY OPERATIONS")
print("=" * 60)

print("Sum of All Elements:", np.sum(arr2))
print("Sum of Each Row:", np.sum(arr2, axis=1))
print("Sum of Each Column:", np.sum(arr2, axis=0))
print("Mean:", np.mean(arr2))
print("Maximum:", np.max(arr2))
print("Minimum:", np.min(arr2))
print("Standard Deviation:", np.std(arr2))
print("Transpose:\n", arr2.T)

# ============================================================
# 4. PANDAS DATAFRAME
# ============================================================

print("\n" + "=" * 60)
print("4. PANDAS DATAFRAME")
print("=" * 60)

data = {
    "Name": ["Asha", "Bhavya", "Chhavi", "Diya"],
    "Age": [19, 20, 19, 21],
    "Marks": [85, 92, 78, 95],
    "Department": ["AI&ML", "CSE", "AI&ML", "ISE"]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Pandas indexing
print("\n--- Pandas Indexing Operations ---")
print("First Five Rows:\n", df.head())
print("Last Two Rows:\n", df.tail(2))
print("Name Column:\n", df["Name"])
print("First Two Rows:\n", df.iloc[:2])
print("First Two Columns:\n", df.iloc[:, :2])
print("Student at Row 1, Column 2:", df.iloc[1, 2])

# Filtering and statistical operations
print("\n--- Pandas Filtering and Operations ---")
print("Students with Marks Greater Than 80:")
print(df[df["Marks"] > 80])

print("Average Marks:", df["Marks"].mean())
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())

print("\nDataFrame Sorted by Marks:")
print(df.sort_values(by="Marks", ascending=False))

# ============================================================
# 6. FINAL DATAFRAME OPERATIONS
# ============================================================

print("\n" + "=" * 60)
print("6. FINAL DATAFRAME OPERATIONS")
print("=" * 60)

print("DataFrame Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe())

print("\nColumn Names:")
print(df.columns)

print("\nDataFrame Shape:")
print(df.shape)
```
